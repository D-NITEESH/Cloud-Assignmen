# Cloud Computing (CS3081) Mini Project: Teacher Labelling & Student Distillation

This Kaggle notebook runs the **heavy compute** steps of your assignment on free GPU (T4 x2 or P100):
1. **Teacher Labelling**: Uses `microsoft/Phi-3.5-mini-instruct` (same architecture & weights as `phi3.5`) to label all 3,000 commands in `commands_pool.jsonl` with greedy decoding (`temperature=0, seed=42`).
2. **Sharding**: Outputs 4 separate shards (`shard-0.jsonl` through `shard-3.jsonl`).
3. **Data Cleaning (`clean.py`)**: Validates schema rules (a)-(e), checks the counting invariant, generates `datasets/v1/train.jsonl` and `train.sha256`.
4. **Fine-Tuning Pythia-70M**: Trains `models/v1` and `models/v2` (sequence-level distillation) and exports `loss.csv`.
5. **Evaluation (`score.py`)**: Evaluates Teacher, Untrained Base, and Distilled Student on `validation.jsonl`.
6. **Release Packaging**: Generates tarballs with SHA-256 checksums and a single download zip `cloud_assignment_artifacts.zip`.
